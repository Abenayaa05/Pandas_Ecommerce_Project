## Phase_04 - Data Transformation and Management

```python
#31 Create a working copy of the original DataFrame and retain the untouched source as df_raw.

df_raw = df.copy(deep=True)
df = df_raw.copy(deep=True)
```

```python
#32 Rename the Description column to ProductDescription in your working copy.

df.rename(
    columns={'Description': 'ProductDescription'},
    inplace=True
)
```

```python
#33 Create a Revenue column by multiplying Quantity by UnitPrice, preserving the original signed quantity.

df['Revenue'] = df['Quantity'] * df['UnitPrice']
```

```python
#34	Create a TransactionType column to classify records as cancellation, return, positive sale or other/invalid, using documented rules.

df['TransactionType'] = np.select(
    [
        df['InvoiceNo'].str.startswith('C', na=False),
        df['Quantity'] < 0,
        (df['Quantity'] > 0) & (df['UnitPrice'] > 0)
    ],
    [
        'cancellation',
        'return',
        'positive sale'
    ],
    default='other/invalid'
)
```

```python
#35 Create an IsIdentifiedCustomer column that indicates whether CustomerID is available.

df['IsIdentifiedCustomer'] = df['CustomerID'].notna()
```

```python
#36 Convert InvoiceDate to a proper Pandas datetime data type.

df['InvoiceDate'] = pd.to_datetime(
    df['InvoiceDate'],
    errors='coerce'
)
```

```python
#37 Convert InvoiceNo, StockCode and Country to suitable string data types.

for col in ['InvoiceNo', 'StockCode', 'Country']:
    df[col] = df[col].astype('string')
```

```python
#38 Reorder columns so transaction identifiers and dates appear before product, customer and financial information.

column_order = [
    'InvoiceNo',
    'InvoiceDate',
    'StockCode',
    'ProductDescription',
    'Country',
    'CustomerID',
    'Quantity',
    'UnitPrice',
    'Revenue',
    'TransactionType',
    'IsIdentifiedCustomer'
]
remaining_columns = [
    col for col in df.columns if col not in column_order
]
df = df[column_order + remaining_columns]
```

```python
#39 Sort the working dataset by InvoiceDate and then by InvoiceNo.

df = df.sort_values(
    by=['InvoiceDate', 'InvoiceNo'],
    ascending=[True, True],
    na_position='last'
)
```

```python
#40 Reset the DataFrame index after sorting and verify that the new index is sequential.

df = df.reset_index(drop=True)

print(df.head())
print("\nDataFrame shape:", df.shape)
print("\nData types:\n", df.dtypes)
print("\nIs index sequential?", df.index.equals(pd.RangeIndex(len(df))))
```

## Phase_05 - Data Quality and Missing-Value Management

```python
#41 Calculate the missing-value count and missing-value percentage for every column.

df_raw = df.copy(deep=True)
df_work = df_raw.copy(deep=True)

missing_count = df_work.isnull().sum()
missing_percentage = (missing_count / len(df_work) * 100).round(2)

missing_summary = pd.DataFrame({
    'Missing_Count': missing_count,
    'Missing_Percentage': missing_percentage
})
print("1. Missing Values:")
print(missing_summary)
```

```python
#42	Identify the columns with the highest percentage of missing values and rank them.

missing_ranking = missing_summary.sort_values(
    by='Missing_Percentage',
    ascending=False
)

print("\n2. Missing-Value Ranking:")
print(missing_ranking)
```

```python
#43	Display all records where CustomerID is missing and determine how many transactions are affected.

missing_customer = df_work[df_work['CustomerID'].isna()]
print("\n3. Transactions with Missing CustomerID:")
print(missing_customer)
print("Affected transactions:", len(missing_customer))
```

```python
#44	Display records where Description is missing or contains only whitespace.

description_text = df_work['ProductDescription'].astype('string')

missing_description_mask = (
    description_text.isna()
    | description_text.str.strip().eq('').fillna(False)
)
missing_description = df_work[missing_description_mask]
print("\n4. Records with Missing or Blank Description:")
print(missing_description)
print("Affected records:", len(missing_description))
```

```python
#45	Create a missing-value report with column name, missing count, non-missing count and missing percentage.

missing_report = pd.DataFrame({
    'Column_Name': df_work.columns,
    'Missing_Count': df_work.isna().sum().values,
    'Non_Missing_Count': df_work.notna().sum().values,
    'Missing_Percentage': (
        df_work.isna().sum().values / len(df_work) * 100
    ).round(2)
})
print("\n5. Missing-Value Report:")
print(missing_report)
```

```python
#46	Identify exact duplicate rows and report the duplicate count before removing anything.

duplicate_mask = df_work.duplicated(keep=False)
exact_duplicate_rows = df_work[duplicate_mask]

duplicate_count = df_work.duplicated().sum()

print("\n6. Exact Duplicate Rows:")
print(exact_duplicate_rows)
print("Duplicate rows beyond first occurrences:", duplicate_count)

rows_before = len(df_work)
missing_cells_before = int(df_work.isna().sum().sum())
duplicate_count_before = int(df_work.duplicated().sum())
```

```python
#47	Remove exact duplicate rows from a working copy and verify how many records remain.

df_clean = df_work.drop_duplicates().copy()

print("\n7. Duplicate Removal:")
print("Rows before:", rows_before)
print("Rows after:", len(df_clean))
print("Rows removed:", rows_before - len(df_clean))
```

```python
#48 Investigate repeated InvoiceNo and StockCode combinations without automatically deleting them, since multiple lines may be legitimate.

pair_counts = (
    df_clean.groupby(['InvoiceNo', 'StockCode'], dropna=False)
    .size()
    .reset_index(name='Row_Count')
)
repeated_pairs = pair_counts[pair_counts['Row_Count'] > 1]
print("\n8. Repeated InvoiceNo and StockCode Combinations:")
print(repeated_pairs)
print("Number of repeated combinations:", len(repeated_pairs))
```

```python
#49 Handle missing Description values using a documented strategy and explain why missing CustomerID values should not be filled with invented IDs.

df_clean['ProductDescription'] = df_clean['ProductDescription'].astype('string')

blank_description_mask = (
    df_clean['ProductDescription'].isna()
    | df_clean['ProductDescription'].str.strip().eq('').fillna(False)
)

df_clean.loc[blank_description_mask, 'ProductDescription'] = (
    'Unknown Description'
)

print("\n9. Description Cleaning:")
print(
    "Descriptions replaced:",
    int(blank_description_mask.sum())
)
print(
    "Missing CustomerIDs remaining:",
    int(df_clean['CustomerID'].isna().sum())
)
```

```python
#50 Create a data-quality report comparing row counts, missing values and duplicate counts before and after cleaning.

quality_report = pd.DataFrame({
    'Metric': [
        'Total Rows',
        'Total Columns',
        'Missing Cells',
        'Rows with Missing CustomerID',
        'Rows with Missing or Blank ProductDescription',
        'Exact Duplicate Rows Beyond First Occurrences'
    ],
    'Before_Cleaning': [
        rows_before,
        df_work.shape[1],
        missing_cells_before,
        int(df_work['CustomerID'].isna().sum()),
        int(missing_description_mask.sum()),
        duplicate_count_before
    ],
    'After_Cleaning': [
        len(df_clean),
        df_clean.shape[1],
        int(df_clean.isna().sum().sum()),
        int(df_clean['CustomerID'].isna().sum()),
        int(
            (
                df_clean['ProductDescription'].isna()
                | df_clean['ProductDescription']
                    .str.strip().eq('').fillna(False)
            ).sum()
        ),
        int(df_clean.duplicated().sum())
    ]
})

print("\n10. Before-and-After Data Quality Report:")
print(quality_report.to_string(index=False))

missing_report.to_csv('missing_value_report.csv', index=False)
quality_report.to_csv('data_quality_report.csv', index=False)

print("\nReports exported successfully.")
```
